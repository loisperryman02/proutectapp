```python
from ctransformers import AutoModelForCausalLM
```

Mistral 7B - small model 
Quantization - compresses the parameters of a model to ensure that less RAM and less VRAM is used in the model. 
Load original model, then quantize it - quite computationally expensive.

TheBloke - already quantized model. 

```python
model = AutoModelForCausalLM.from_pretrained(
    "TheBloke/Mistral-7B-Instruct-v0.1-GGUF",
    model_file="mistral-7b-instruct-v0.1.Q4_K_M.gguf",
    model_type="mistral",
    gpu_layers=0,
    hf=True
)
```

```python
from transformers import AutoTokenizer, pipeline

# Tokenizer
tokenizer = AutoTokenizer.from_pretrained("mistralai/Mistral-7B-Instruct-v0.1")

generator = pipeline(
    model=model, tokenizer=tokenizer,
    task = "text-generation",
    max_new_tokens = 50,
    repetition_penalty = 1.1
)



```
