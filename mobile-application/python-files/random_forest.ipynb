```python
training_set = pd.read_csv("C:\\Users\\loisp\\Documents\\Year3\\Dissertation\\Proutect\\Proutect\\csv-json-files\\training_set.csv")


features = training_set.iloc[:,0:4].values
target_variables = training_set.iloc[:,5].values

training_set.info()

print(target_variables)
```

The following code processes categorical data by numericlaly encoding it and combines the processed data with numerical data. The Random Forest Regression model is then trained with the prepared data. 

```python
label_encoder = LabelEncoder()
x = training_set.select_dtypes(exclude=['object']).values

print(x)
# Fitting Random Forest Regression to the dataset
regressor = RandomForestRegressor(n_estimators=10, random_state=0, oob_score=True)

# Fit the regressor with x and y data
regressor.fit(x, target_variables)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

oob_score = regressor.oob_score_
print(f'Out-of-Bag Score: {oob_score}')
 
# Making predictions on the same data or new data
predictions = regressor.predict(x)
 
# Evaluating the model
mse = mean_squared_error(target_variables, predictions)
print(f'Mean Squared Error: {mse}')
 
r2 = r2_score(target_variables, predictions)
print(f'R-squared: {r2}')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor

# Assuming your Random Forest model is named 'regressor'
# and it was trained on a dataset with 5 features

# Let's say features is a 2D numpy array of shape (n_samples, 5)
# and target_variable is a 1D numpy array of shape (n_samples,)

num_features = features.shape[1]  # Number of features
num_points = 100  # Number of points in the grid for plotting

# Calculate the mean of each feature
mean_features = np.mean(features, axis=0)

# Prepare the plot
plt.figure(figsize=(15, 10))

for i in range(num_features):
    # Create a grid for the current feature
    feature_range = np.linspace(np.min(features[:, i]), np.max(features[:, i]), num_points)
    
    # Create a dataset for predictions where we vary the current feature
    # while keeping other features fixed at their mean values
    features_for_prediction = np.tile(mean_features, (num_points, 1))
    features_for_prediction[:, i] = feature_range

    # Make predictions
    predictions = regressor.predict(features_for_prediction)

    # Plotting
    plt.subplot(3, 2, i + 1)
    plt.scatter(features[:, i], target_variables, color='blue')
    plt.plot(feature_range, predictions, color='green')
    plt.title(f"Feature {i+1} vs Target")
    plt.xlabel(f"Feature {i+1}")
    plt.ylabel('Target Variable')

plt.tight_layout()
plt.show()
```
